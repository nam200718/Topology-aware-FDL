```python
import yaml
import os

with open('temp/baseline_comparison.yaml', 'r') as f:
    config = yaml.safe_load(f)

# RÚT GỌN CHO NHANH:
config['num_rounds'] = 2          # Đổi từ 50 xuống 2 round
config['clients']['local_steps'] = 1 # Đổi từ 2 xuống 1 step
config['env']['output_dir'] = './outputs/idea3_fast_debug' # Thay đổi thư mục lưu

with open('temp/baseline_comparison_fast.yaml', 'w') as f:
    yaml.dump(config, f)

!python temp/run_baseline_comparison.py --config temp/baseline_comparison_fast.yaml

# Lưu kết quả riêng biệt
import shutil
from IPython.display import FileLink, display
shutil.make_archive('idea3_fast_debug', 'zip', 'outputs/idea3_fast_debug')
print('Tải file Idea 3 (Fast Debug):')
display(FileLink('idea3_fast_debug.zip'))
```

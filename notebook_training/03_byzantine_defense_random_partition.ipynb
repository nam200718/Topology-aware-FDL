## Run Experiment (50 rounds, batch_size=128, label_flip)


```python
import yaml, os

# 1. Define dedicated output directory (isolated from repo)
output_dir = '/kaggle/working/output_partB_E4_label_flip' if os.path.exists('/kaggle/working') else './outputs/output_partB_E4_label_flip'
os.makedirs(output_dir, exist_ok=True)

# 2. Load base config and override
with open('temp/exp4_defense_random.yaml', 'r') as f:
    cfg = yaml.safe_load(f)

cfg['byzantine_types'] = ['label_flip']
cfg['env']['output_dir'] = output_dir

run_yaml = 'temp/exp4_defense_random_run.yaml'
with open(run_yaml, 'w') as f:
    yaml.dump(cfg, f)

# 3. Run experiment
!python temp/run_baseline_comparison.py --config temp/exp4_defense_random_run.yaml
```
